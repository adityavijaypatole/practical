```python
import pandas as pd
import numpy as np

mock_re = {
    ' Property Type ': ['House', 'Condo', 'House', 'Apartment', 'House', 'Condo'],
    'Sale_Price_$': [250000, 150000, np.nan, 100000, 5000000, 160000], # np.nan is missing, 5M is an outlier
    'Location_ID': [1, 2, 1, 3, 2, 1]
}
pd.DataFrame(mock_re).to_csv('RealEstate_Prices.csv', index=False)

mock_demo = {
    'Location_ID': [1, 2, 3],
    'Safety_Rating': ['High', 'Medium', 'Low']
}
pd.DataFrame(mock_demo).to_csv('Neighborhood_Demographics.csv', index=False)
```

```python
import pandas as pd
import numpy as np

df = pd.read_csv('RealEstate_Prices.csv')

df.columns = df.columns.str.strip().str.replace(' ', '_').str.replace('$', '')
print("Cleaned Columns:", df.columns.tolist())

median_price = df['Sale_Price_'].median()
df['Sale_Price_'] = df['Sale_Price_'].fillna(median_price)

df_demo = pd.read_csv('Neighborhood_Demographics.csv')

df = pd.merge(df, df_demo, on='Location_ID', how='left')

df = df[df['Sale_Price_'] < 1000000]

location_1_data = df[df['Location_ID'] == 1]
print(f"\nFiltered Data (Only Location 1):\n{location_1_data}")

df = pd.get_dummies(df, columns=['Property_Type', 'Safety_Rating'])

avg_price_by_location = df.groupby('Location_ID')['Sale_Price_'].mean().reset_index()
print("\nAverage Sale Price by Location:\n", avg_price_by_location)
```
