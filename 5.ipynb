```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
```

```python
col_names = ['buying', 'maint', 'doors', 'persons', 'lug_boot', 'safety', 'target_class']
df = pd.read_csv('car_evaluation.csv', names=col_names)

X = pd.get_dummies(df.drop('target_class', axis = 1))
Y = df['target_class']

x_train, x_test, y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state = 42) 

rf_model = RandomForestClassifier(n_estimators = 100, random_state = 42)
rf_model.fit(x_train,y_train)

y_pred = rf_model.predict(x_test)

print(f"Random Forest Accuracy: {accuracy_score(y_test, y_pred)*100:.2f}")
print("Classification Report:\n\n", classification_report(y_test, y_pred))
```
