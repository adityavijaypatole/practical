```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis as LDA
```

```python
# url = 'https://www.kaggle.com/datasets/uciml/iris'
# iris_df = pd.read_csv(url)

# iris_df = pd.read_csv('Iris.csv')

iris_df = sns.load_dataset('iris')
```

```python
# The Kaggle Iris CSV usually has an 'Id' column at index 0. 
# We use iloc to skip column 0, take columns 1 through 4 for X, and the last column for y.
X = iris_df.iloc[:, 1:-1]
Y = iris_df.iloc[:, -1]

# We compress the 4 flower measurements down into 2 Linear Discriminants.
lda = LDA(n_components = 2)

X_lda = lda.fit_transform(X,Y)

plt.figure(figsize = (8,5))
sns.scatterplot(x = X_lda[:, 0], y = X_lda[:,1], hue = Y, palette = 'bright')# can add hue=y,palette = 'bright' and stuff for visulaization

plt.title('LDA discriminatory Reduction on Iris Dataset')
plt.xlabel('Linear Discriminant 1')
plt.ylabel('Linear Discriminant 2')
plt.show()
```
