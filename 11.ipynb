```python
import pandas as pd

mock_aqi = {
    'Date': ['2023-11-01', '2023-11-02', '2023-11-03', '2023-11-04', '2023-11-05'],
    'PM2.5': [35, 42, 55, 30, 20],
    'PM10': [60, 75, 90, 50, 40],
    'CO': [0.5, 0.7, 0.9, 0.4, 0.3],
    'AQI': [110, 130, 160, 95, 75]
}
pd.DataFrame(mock_aqi).to_csv('City_Air_Quality.csv', index=False)
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('City_Air_Quality.csv')
print("Dataset Structure:\n", df.head())

df['Date'] = pd.to_datetime(df['Date'])

plt.figure(figsize=(14, 10))

plt.subplot(2, 2, 1)

plt.plot(df['Date'], df['AQI'], label='AQI', color='red', marker='o', linewidth=2)
plt.plot(df['Date'], df['PM2.5'], label='PM2.5', color='blue', linestyle='--')
plt.plot(df['Date'], df['PM10'], label='PM10', color='green', linestyle=':')
plt.title('AQI and Pollutant Trends Over Time (Task 4 & 5)')
plt.xticks(rotation=45)
plt.legend()

plt.subplot(2, 2, 2)

date_strings = df['Date'].dt.strftime('%Y-%m-%d')
plt.bar(date_strings, df['AQI'], color='orange', edgecolor='black')
plt.title('AQI Values by Date (Task 6)')
plt.xticks(rotation=45)
plt.ylabel('Air Quality Index')

plt.subplot(2, 2, 3)

plt.boxplot([df['PM2.5'], df['PM10'], df['CO']], tick_labels=['PM2.5', 'PM10', 'CO'])
plt.title('Distribution of Pollutant Levels (Task 7)')
plt.ylabel('Concentration')

plt.subplot(2, 2, 4)
plt.scatter(df['PM2.5'], df['AQI'], color='purple', s=100, alpha=0.7)
plt.title('Correlation: AQI vs PM2.5 (Task 8)')
plt.xlabel('PM2.5 Levels')
plt.ylabel('AQI')

plt.tight_layout()
plt.show()
```
